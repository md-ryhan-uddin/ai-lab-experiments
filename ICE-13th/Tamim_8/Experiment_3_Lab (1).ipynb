# Production rules represented using Python List of Dictionaries

```python
# Production rules represented using Python List of Dictionaries

rules = [
    {
        "conditions": ["has_fur"],
        "conclusion": "mammal"
    },
    {
        "conditions": ["has_scales"],
        "conclusion": "reptile"
    },
    {
        "conditions": ["mammal", "can_walk"],
        "conclusion": "walking_mammal"
    },
    {
        "conditions": ["mammal", "cannot_walk"],
        "conclusion": "non_walking_mammal"
    }
]

print(CYAN + "Production Rules" + RESET)
pprint(rules)
```

# Forward Chaining Algorithm Implementation

```python
# Forward Chaining Algorithm Implementation

def forward_chaining(facts, rules):
    inferred = set()
    changed = True

    while changed:
        changed = False

        for rule in rules:
            if all(
                facts.get(condition, False) or condition in inferred
                for condition in rule["conditions"]
            ):
                conclusion = rule["conclusion"]

                if conclusion not in inferred:
                    inferred.add(conclusion)
                    changed = True

    return inferred
```

# Test the forward chaining algorithm

```python
# Test the forward chaining algorithm

facts = {
    "has_fur": True,
    "can_walk": True
}

result = forward_chaining(facts, rules)

print(BLUE + "Inferred Knowledge" + RESET)
print(GREEN + str(result) + RESET)
```

```python
facts = {
    "has_fur": True,
    "cannot_walk": True
}

result = forward_chaining(facts, rules)

print(YELLOW + "New Inferred Knowledge" + RESET)
print(CYAN + str(result) + RESET)
```

# Disease Diagnosis Expert System represented using Python List of Dictionaries

```python
# Disease Diagnosis Expert System represented using Python List of Dictionaries

disease_rules = [

    {
        "conditions": ["fever", "headache"],
        "disease": "Common Cold"
    },
    {
        "conditions": ["fever", "rash"],
        "disease": "Chickenpox"
    },
    {
        "conditions": ["headache", "nausea"],
        "disease": "Food Poisoning"
    }
]
```

# Function to diagnose disease based on symptoms

```python
# Function to diagnose disease based on symptoms

def diagnose(symptoms):
    possible = []

    for rule in disease_rules:
        if all(symptom in symptoms for symptom in rule["conditions"]):
            possible.append(rule["disease"])

    return possible
```

# Example usage of the diagnosis function

```python
# Example usage of the diagnosis function

patient = [
    "fever",
    "headache"
]

diagnosis = diagnose(patient)

print(CYAN + "Symptoms:" + RESET, YELLOW + str(patient) + RESET)
print(GREEN + "Diagnosis:" + RESET, RED + str(diagnosis) + RESET)
```

# User Input for Symptoms

```python
# User Input for Symptoms

symptoms = []

questions = [
    "fever",
    "headache",
    "rash",
    "nausea",
    "cough"
]

print(BLUE + "Answer using y/n" + RESET)

for question in questions:
    ans = input(YELLOW + f"Do you have {question}? " + RESET)

    if ans.lower() == "y":
        symptoms.append(question)

result = diagnose(symptoms)

print("\n" + CYAN + "Possible Disease" + RESET)

if result:
    print(GREEN + str(result) + RESET)
else:
    print(RED + "No disease matched." + RESET)
```

```python
def explain(symptoms):

    print(BLUE + "Reasoning Process" + RESET)

    for rule in disease_rules:

        if all(symptom in symptoms for symptom in rule["conditions"]):

            print()

            print(GREEN + "Matched Rule" + RESET)

            print(
                YELLOW + "IF " +
                " AND ".join(rule["conditions"]) +
                RESET
            )

            print(
                CYAN + "THEN " +
                rule["disease"] +
                RESET
            )
```

```python
patient = [
    "fever",
    "headache"
]

diagnosis = diagnose(patient)

print(GREEN + "Diagnosis:" + RESET, diagnosis)

explain(patient)
```

```python
knowledge = {
    "mammal": ["has_fur"],
    "walking_mammal": ["mammal", "can_walk"],
    "non_walking_mammal": ["mammal", "cannot_walk"]
}
```

```python
def backward_chain(goal, facts):

    if goal in facts:
        return True

    if goal not in knowledge:
        return False

    for condition in knowledge[goal]:

        if not backward_chain(condition, facts):
            return False

    return True
```

```python
facts = {
    "has_fur",
    "can_walk"
}

goal = "walking_mammal"

print(BLUE + "Goal:" + RESET, YELLOW + goal + RESET)
print(CYAN + "Can be proved?" + RESET)

result = backward_chain(goal, facts)

if result:
    print(GREEN + "True" + RESET)
else:
    print(RED + "False" + RESET)
```
