### Mathematical Background

In AI, a dataset is often represented as a matrix. If there are `n` samples and `d` features, the input data can be written as:

$$X \in \mathbb{R}^{n \times d}$$

Each row of `X` represents one observation, and each column represents one feature. NumPy makes it easy to perform vectorized operations such as:

$$y = Xw + b$$

Before training a model, we often calculate basic statistics. The mean and variance of a feature are:

$$\mu = \frac{1}{n}\sum_{i=1}^{n}x_i$$

$$\sigma^2 = \frac{1}{n}\sum_{i=1}^{n}(x_i - \mu)^2$$

These values help us understand the scale, spread, and distribution of the data.

````python
%env MPLBACKEND cairo
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('cairo')
import matplotlib.pyplot as plt
from sklearn.datasets import load_linnerud
from IPython.display import display

try:
    import ipywidgets as widgets
    from ipywidgets import interact
    widgets_available = True
except ImportError:
    widgets_available = False

np.random.seed(2026)
plt.style.use("seaborn-v0_8-whitegrid")

print("Setup complete. All libraries loaded successfully.")
````

````python
numbers = [2, 4, 6, 8, 10]

squared_numbers = [value ** 2 for value in numbers]
cubed_numbers = [value ** 3 for value in numbers]

display(pd.DataFrame({
    "number": numbers,
    "square": squared_numbers,
    "cube": cubed_numbers,
}))
````

````python
vector = np.array(numbers)

multiplier = 3
bias = 5
scaled_vector = vector * multiplier + bias

print("Original vector:", vector)
print("Scaled vector:", scaled_vector)
````

````python
matrix = np.arange(10, 19).reshape(3, 3)
print("\nMatrix:")
display(pd.DataFrame(matrix))

print("Row means:", matrix.mean(axis=1))
````

````python
print("\nBroadcast addition:")
display(pd.DataFrame(matrix + np.array([100, 200, 300])))
````

````python
linnerud = load_linnerud()
exercise_df = pd.DataFrame(linnerud.data, columns=linnerud.feature_names)
physiology_df = pd.DataFrame(linnerud.target, columns=linnerud.target_names)
linnerud_df = pd.concat([exercise_df, physiology_df], axis=1)

print("Dataset loaded successfully.")
print("Shape:", linnerud_df.shape)
print("Columns:", list(linnerud_df.columns))
print("Number of samples: 20")

if widgets_available:
    def show_rows(n=5):
        display(linnerud_df.head(n))

    interact(show_rows, n=widgets.IntSlider(value=5, min=1, max=20, step=1))
````

````python
missing_values = linnerud_df.isna().sum()

print("Missing values per column:")
display(missing_values.to_frame("missing_count"))
````

````python
print("Descriptive statistics:")
display(linnerud_df.describe(include="all"))
````

````python
correlations = linnerud_df.corr()
print("Feature correlations:")
display(correlations.round(3))

print()
print("Strongest correlation between exercise and physiology:")
exercise_cols = linnerud.feature_names
physiology_cols = linnerud.target_names
for ex in exercise_cols:
    for ph in physiology_cols:
        val = correlations.loc[ex, ph]
        if abs(val) > 0.4:
            print(f"  {ex} vs {ph}: {val:.3f}")
````

````python
if widgets_available:
    def compare_features(f1=linnerud.feature_names[0], f2=linnerud.target_names[0]):
        plt.figure(figsize=(6, 4))
        plt.scatter(linnerud_df[f1], linnerud_df[f2], color="#2a9d8f", s=60)
        plt.xlabel(f1)
        plt.ylabel(f2)
        plt.title(f"{f1} vs {f2}")
        plt.tight_layout()
        plt.show()

    interact(compare_features,
             f1=list(linnerud.feature_names),
             f2=list(linnerud.target_names))
````

````python
numeric_columns = linnerud_df.columns

if widgets_available:
    def plot_histogram(feature=numeric_columns[0], bins=15):
        plt.figure(figsize=(7, 4))
        plt.hist(linnerud_df[feature], bins=bins, color="#e9c46a", edgecolor="black")
        plt.title(f"Distribution of {feature}")
        plt.xlabel(feature)
        plt.ylabel("Count")
        plt.tight_layout()
        plt.show()

    interact(
        plot_histogram,
        feature=numeric_columns,
        bins=widgets.IntSlider(value=15, min=5, max=30, step=1),
    )
else:
    fig, axes = plt.subplots(2, 3, figsize=(12, 7))
    axes = axes.ravel()
    for axis, column in zip(axes.ravel(), numeric_columns):
        axis.hist(linnerud_df[column], bins=15, color="#e9c46a", edgecolor="black")
        axis.set_title(column)
    plt.tight_layout()
    plt.show()
````

````python
all_columns = list(linnerud_df.columns)

def scatter_features(x_feature="Chins", y_feature="Weight"):
    plt.figure(figsize=(7, 5))
    plt.scatter(
        linnerud_df[x_feature],
        linnerud_df[y_feature],
        color="#2a9d8f",
        alpha=0.8,
        s=60,
    )
    plt.xlabel(x_feature)
    plt.ylabel(y_feature)
    plt.title(f"{x_feature} vs {y_feature}")
    plt.tight_layout()
    plt.show()

if widgets_available:
    interact(scatter_features, x_feature=all_columns, y_feature=all_columns)
else:
    scatter_features()
````

````python
correlation_matrix = linnerud_df.corr()

plt.figure(figsize=(7, 5))
plt.imshow(correlation_matrix, cmap="plasma", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")
plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns, rotation=45, ha="right")
plt.yticks(range(len(correlation_matrix.index)), correlation_matrix.index)

for row in range(len(correlation_matrix.index)):
    for col in range(len(correlation_matrix.columns)):
        plt.text(col, row, f"{correlation_matrix.iloc[row, col]:.2f}", ha="center", va="center", color="white")

plt.title("Feature correlation heatmap")
plt.tight_layout()
plt.show()
````

````python
summary_table = pd.DataFrame({
    "mean": linnerud_df.mean(),
    "std": linnerud_df.std(),
    "min": linnerud_df.min(),
    "max": linnerud_df.max(),
})

display(summary_table.round(3))
````
